```python
def dominant_frequency(signal, dt):
    quitarmedia = signal - signal.mean()

    fft_values = np.fft.rfft(quitarmedia)

    frequencies = np.fft.rfftfreq(len(quitarmedia), d=dt)

    amplitude = np.abs(fft_values)

    dominant_index = np.argmax(amplitude[1:]) + 1

    dominant_frequency = frequencies[dominant_index]

    return dominant_frequency
```

```python
freq_roll = dominant_frequency(oscillation["roll_deg"], dt)
freq_rudder = dominant_frequency(oscillation["rudder_deg"], dt)
freq_yaw = dominant_frequency(oscillation["yaw_rate_dps"], dt)

freq_roll, freq_rudder, freq_yaw
```

```python
roll_pp = oscillation["roll_deg"].max() - oscillation["roll_deg"].min()
roll_pp
```

```python
rudder_pp = oscillation["rudder_deg"].max() - oscillation["rudder_deg"].min()
rudder_pp
```

```python
yawrate_pp = oscillation["yaw_rate_dps"].max() - oscillation["yaw_rate_dps"].min()
yawrate_pp
```

```python
rudder_norm = (
    oscillation["rudder_deg"] - oscillation["rudder_deg"].mean()
) / oscillation["rudder_deg"].std()

yaw_norm = (
    oscillation["yaw_rate_dps"] - oscillation["yaw_rate_dps"].mean()
) / oscillation["yaw_rate_dps"].std()
```

```python
corr = np.correlate(rudder_norm, yaw_norm, mode="full")

lags = np.arange(-len(rudder_norm) + 1, len(rudder_norm))
```

```python
best_lag = lags[np.argmax(corr)]

lag_time = best_lag * dt

best_lag, lag_time
```

```python
roll_norm = (
    oscillation["roll_deg"] - oscillation["roll_deg"].mean()
) / oscillation["roll_deg"].std()
```

```python
corr = np.correlate(roll_norm, yaw_norm, mode="full")

lags = np.arange(-len(roll_norm) + 1, len(roll_norm))
```

```python
best_lag = lags[np.argmax(corr)]

lag_time = best_lag * dt

best_lag, lag_time
```

```python
plt.plot(oscillation["time_s"], yaw_norm, label="Yaw rate")
plt.plot(oscillation["time_s"], roll_norm, label="Roll")

plt.xlabel("Time [s]")
plt.ylabel("Normalized amplitude")
plt.title("Yaw rate vs Roll")
plt.legend()
plt.grid()
plt.show()
```
