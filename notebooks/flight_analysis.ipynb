```python
oscillation = test_point[
    (test_point["time_s"] >= 128) &
    (test_point["time_s"] <= 143)
]
```

```python
oscillation.shape
```

```python
plt.plot(oscillation["time_s"], oscillation["pitch_deg"])

plt.xlabel("Time [s]")
plt.ylabel("Pitch [deg]")
plt.title("Pitch Oscillation")
plt.grid()
plt.show()
```

```python
import numpy as np
```

```python
dt = np.mean(np.diff(oscillation["time_s"]))
fs = 1 / dt

dt, fs
```

```python
signal = oscillation["pitch_deg"] - oscillation["pitch_deg"].mean()

fft_values = np.fft.rfft(signal)
frequencies = np.fft.rfftfreq(len(signal), d=dt)

amplitude = np.abs(fft_values)
```

```python
dominant_index = np.argmax(amplitude[1:]) + 1
dominant_frequency = frequencies[dominant_index]

dominant_frequency
```

```python
elevator_signal = oscillation["elevator_deg"] - oscillation["elevator_deg"].mean()

elevator_fft = np.fft.rfft(elevator_signal)
elevator_amplitude = np.abs(elevator_fft)

elevator_dominant_index = np.argmax(elevator_amplitude[1:]) + 1
elevator_dominant_frequency = frequencies[elevator_dominant_index]

elevator_dominant_frequency
```

```python
pitch_norm = (
    oscillation["pitch_deg"] - oscillation["pitch_deg"].mean()
) / oscillation["pitch_deg"].std()

elevator_norm = (
    oscillation["elevator_deg"] - oscillation["elevator_deg"].mean()
) / oscillation["elevator_deg"].std()

plt.plot(oscillation["time_s"], pitch_norm, label="Pitch")
plt.plot(oscillation["time_s"], elevator_norm, label="Elevator")

plt.xlabel("Time [s]")
plt.ylabel("Normalized amplitude")
plt.title("Pitch vs Elevator")
plt.legend()
plt.grid()
plt.show()
```

```python
correlation = np.corrcoef(pitch_norm, elevator_norm)[0, 1]
correlation
```

```python
corr = np.correlate(pitch_norm, elevator_norm, mode="full")

lags = np.arange(-len(pitch_norm) + 1, len(pitch_norm))

best_lag = lags[np.argmax(corr)]

lag_time = best_lag * dt

best_lag, lag_time
```

```python
max_lag_samples = int((0.5 / dominant_frequency) / dt)

center = len(corr) // 2

corr_small = corr[
    center - max_lag_samples :
    center + max_lag_samples + 1
]

lags_small = lags[
    center - max_lag_samples :
    center + max_lag_samples + 1
]

best_lag_small = lags_small[np.argmax(corr_small)]
lag_time_small = best_lag_small * dt

best_lag_small, lag_time_small
```
